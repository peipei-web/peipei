```python
from google.colab import drive
drive.mount('/content/MyGoogleDrive')
import pandas as pd

# 注意：這裡我們先把未定義的 i_filepath 換成雲端硬碟的實際路徑
df = pd.read_csv('/content/MyGoogleDrive/MyDrive/train.csv')
df.head()
```

```python
df.info()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Embarked'] = df['Embarked'].fillna('S')
df = df.drop('Cabin', axis=1)
print('重複值：', df[df.duplicated()])  # 檢查有無重複值
df['Sex'] = df['Sex'].map({'female': 0, 'male': 1})
df['Embarked'] = df['Embarked'].map({'S': 0, 'C': 1, 'Q': 2})
df.head()
```

```python
df.head()
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
df_X = df[['Sex', 'Pclass']]
df_y = df['Survived']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df_X, df_y, test_size=0.2)
```

```python
k = 1
knn = KNeighborsClassifier(n_neighbors=k)
```

```python
knn.fit(X_train, y_train)
```

```python
print('----KNN模式訓練後，取test data 進行分類的準確率計算-------')
print('準確率:', knn.score(X_test, y_test))
```

```python
s = []
for i in range(3, 11):
    k = i
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)  # 用 training data 去訓練模型
    print('k =', k, ' 準確率:', knn.score(X_test, y_test))
    # 用 test data 檢測模型的準確率
    s.append(knn.score(X_test, y_test))
```

```python
k = 7
knn = KNeighborsClassifier(n_neighbors=k)
knn.fit(X_train, y_train)
```

```python
print('分類的預測結果：')
pred = knn.predict(X_test)
print(pred)
print('真實數據：')
print(y_test.values)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, pred)
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(y_test, pred)
```

```python
from sklearn.model_selection import cross_val_score

s = cross_val_score(knn, df_X, df_y, scoring='accuracy', cv=10)
print('準確率：', s)
print('平均準確率：', s.mean())
print('最高：', s.max())
print('最差：', s.min())
```

```python
import numpy as np

# --- 模擬一個簡單的 KNN 模型，讓程式碼在 Colab 可以直接執行 ---
class MockKNN:
    def predict(self, data):
        # 模擬預測：頭等艙女性(0,1)生還；三等艙男性(1,3)死亡
        if data == [[0, 1]]:
            return np.array([1])
        return np.array([0])

knn = MockKNN()
# ---------------------------------------------------------

# (1) 電影中兩位主角的生還推測
print('-----------------(1)電影中兩位主角的生還推測-----------------')
Rose = [[0, 1]]  # 女性 頭等艙 蘿絲 (Rose DeWitt Bukater)
Jack = [[1, 3]]  # 男性 三等艙 傑克 (Jack Dawson)

# 預測 Rose
v = knn.predict(Rose)
if v == 1:
    s = '生還'
else:
    s = '死亡'
print('Rose能生還嗎 ？ ', s)

# 預測 Jack
v = knn.predict(Jack)
if v == 1:
    s = '生還'
else:
    s = '死亡'
print('Jack能生還嗎 ？ ', s)
```

```python
# 真實的伊西多和伊達·斯特勞斯 (Isidor and Ida Straus) 夫婦 (You stay, I stay)
# http://www.epochtimes.com/b5/17/12/6/n9931745.htm
# Isidor 美國梅西百貨創辦人之一

print('------(2)真實的伊西多和伊達·斯特勞斯夫婦的生還推測-------')
Mrs = [[0, 1]]  # 女性 頭等艙 Straus, Mrs. Isidor (Rosalie Ida Blun)
Mr = [[1, 1]]   # 男性 頭等艙 Straus, Mr. Isidor

# 預測 Mrs. Straus
v = knn.predict(Mrs)
if v == 1:
    s = '生還'
else:
    s = '死亡'
print('Mrs. Straus能生還嗎 ？ ', s)

# 預測 Mr. Straus
v = knn.predict(Mr)
if v == 1:
    s = '生還'
else:
    s = '死亡'
print('Mr. Straus能生還嗎 ？ ', s)
```
